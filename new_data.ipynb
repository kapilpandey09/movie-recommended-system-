```python
from bs4 import BeautifulSoup

# # HTML string provided by the user
# html_string = "<img src='http://image.tmdb.org/t/p/w185//btnl50ZDJDSCal2NLQIYWw0XxvH.jpg' style='height:100px;'>"


def html_url(poster_imgs):


    # Parse the HTML
    L = []
    for i in poster_imgs:
        soup = BeautifulSoup(i, 'html.parser')

    # Find the <img> tag
        img_tag = soup.find('img')

    # Extract the URL from the src attribute
        extracted_url = img_tag['src'] if img_tag else None
        L.append(extracted_url)

    return L
```

```python
poster_url["poster_path"].apply(html_url)
```
